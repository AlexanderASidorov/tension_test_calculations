```python
# импортируем библиотеку для визаулизации данных
import matplotlib.pyplot as plt
import math
```

```python
# Эксперимент 01
stroke_exp_01 = [] # пустой список для записи данных о перемещении
load_exp_01 = [] # пустой список для записи данных о силе
# Эксперимент 02
stroke_exp_02 = [] # пустой список для записи данных о перемещении
load_exp_02 = [] # пустой список для записи данных о силе


# открываем файл с данными и считываем их в списки
with open ('../data/cold_compression_dada/steel_20_exp_01.txt', 'r') as file: 
    for line in file:                                    
        load_stroke = line.split('\t')                  
        stroke_exp_01.append(float(load_stroke[1]))           
        load_exp_01.append(float(load_stroke[2])) 

with open ('../data/cold_compression_dada/steel_20_exp_02.txt', 'r') as file: 
    for line in file:                                    
        load_stroke = line.split('\t')                  
        stroke_exp_02.append(float(load_stroke[1]))           
        load_exp_02.append(float(load_stroke[2])) 

```

```python
# визуализируем результат
# создаем "базу" для график
plt.figure(figsize=(8, 6))
plt.figure(figsize=(8, 6)) # размер графика в дюймах
plt.plot (stroke_exp_01, load_exp_01, '-', color = 'black')
plt.plot (stroke_exp_02, load_exp_02, '--', color = 'black')
# Добавляем названия осей
plt.xlabel ('Ход деформирования, мм', fontsize = 14)
plt.ylabel ('Сила деформирования, Н', fontsize = 14)
# добавляем для красоты сетку
plt.grid()
```

```python
# Размеры образца
d0 = 5 # диаметр образца
L0 = 29.5 # начальная длина образца
A0 = math.pi*(d0**2)/4 # начальная площадь поперечного сечения
```

```python
sigma_eng = [] # пустой список куда будем записывать значения инженерного напряжения
epsilon_eng = [] # пустой список куда будем записывать значения инженерной деформации
for i in range (0, len(load_exp_02)):
    sigma_eng.append(load_exp_02[i]/A0)
    epsilon_eng.append((stroke_exp_02[i]/L0)*100)
```

```python
# Визуализируем результаты
# создаем "базу" для графика
plt.plot (epsilon_eng[10:50], sigma_eng[10:50], '-', color = 'black')
# Добавляем названия осей
plt.xlabel ('Инженерная деформация, %', fontsize = 14)
plt.ylabel ('Инженерное напряжение, МПа', fontsize = 14)
# добавляем для красоты сетку
plt.grid()
```
