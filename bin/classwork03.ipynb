```python
import matplotlib.pyplot as plt
```

```python
eng_strain = []
eng_stress = []
```

```python
%pwd
```

```python
with open ('../data/S550MC.txt', 'r') as file:
    for line in file:
        stress_strain = line.split('\t')
        eng_strain.append(float(stress_strain[0])*100)
        eng_stress.append(float(stress_strain[1]))
```

```python
plt.figure (figsize = (8, 6))
plt.plot(eng_strain, eng_stress)
plt.grid()
```

```python
elastic_limit = 1
elastic_plastic_strain = [0]
elastic_plastic_stress = [0]

for i in range (1, len(eng_strain)):
    if eng_strain[i] < elastic_limit:
        elastic_plastic_strain.append(eng_strain[i])
        elastic_plastic_stress.append(eng_stress[i])
```

```python
plt.figure (figsize = (8, 6))
plt.plot(elastic_plastic_strain, elastic_plastic_stress, marker='o')

plt.grid()
```

```python
E = elastic_plastic_stress[1]/(elastic_plastic_strain[1]/100)
print (f'Модуль Юнга = {E} МПа')
```
